```python
molecules = {
    "erlotinib": "COCCOc1cc2ncnc(Nc3cccc(C#C)c3)c2cc1OCCOC",
    "gefitinib": "COc1cc2ncnc(Nc3ccc(F)c(Cl)c3)c2cc1OCCCN1CCOCC1",
    "aspirin": "CC(=O)OC1=CC=CC=C1C(=O)O"
}
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem

for name, smiles in molecules.items():
  mol = Chem.MolFromSmiles(smiles)
  mol = Chem.AddHs(mol)
  AllChem.EmbedMolecule(mol)
  AllChem.MMFFOptimizeMolecule(mol)
  Chem.MolToPDBFile(mol, f"{name}.pdb")
  print(f"{name} prepared")
```

```python
from vina import Vina
v = Vina(sf_name='vina')
v.set_receptor('1M17_clean.pdb')
center = [22.0, 0.5, 52.0]
box_size = [20, 20, 20]

results = {}
for name in molecules.keys():
    v.set_ligand_from_file(f"{name}.pdb")
    v.compute_vina_maps(center=center, box_size=box_size)
    v.dock(exhaustiveness=8, n_poses=5)
    energy = v.energies()[0][0]
    results[name] = energy
    print(f"{name}: {energy} kcal/mol")
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem
from meeko import MoleculePreparation, PDBQTWriterLegacy

for name, smiles in molecules.items():
    mol = Chem.MolFromSmiles(smiles)
    mol = Chem.AddHs(mol)
    AllChem.EmbedMolecule(mol)
    AllChem.MMFFOptimizeMolecule(mol)

    preparator = MoleculePreparation()
    setups = preparator.prepare(mol)
    pdbqt_string, is_ok, error_msg = PDBQTWriterLegacy.write_string(setups[0])

    with open(f"{name}.pdbqt", "w") as f:
        f.write(pdbqt_string)
    print(f"{name} prepared")
```

```python
import matplotlib.pyplot as plt

names = list(results.keys())
values = list(results.values())

plt.bar(names, values)
plt.ylabel("Binding Affinity (kcal/mol)")
plt.title("Docking Results: EGFR Inhibitors vs Aspirin")
plt.show()
```

**Interpretation**
For the kinase domain of EGFR, the docking of 3 substances was simulated: gefitinib and erlotinib which are known EGFR inhibitors, along with aspirin which played the role of a negative control.

As expected, the known inhibitors showed more negative binding affinities (which is more favourable).  

Exact energy values are more noisy due to this being run at low exhaustion.
